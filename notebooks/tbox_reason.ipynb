# Run pipeline

This notebook runs the flap variant generation pipeline using `wound_ontology` package classes and the native `owlready2` reasoner.

```python
from __future__ import annotations
import os
import sys
from pathlib import Path

NOTEBOOK_DIR = Path.cwd()
REPO_ROOT    = NOTEBOOK_DIR.parent  # notebooks/ → repo root
PACKAGE_ROOT = REPO_ROOT.parents[1] / 'Wound ontology' / 'src'
if str(PACKAGE_ROOT) not in sys.path:
    sys.path.insert(0, str(PACKAGE_ROOT))

from wound_ontology.tbox_abox_reasoner.for_owl.tbox import TBox
from wound_ontology.tbox_abox_reasoner.reasoner.reasoner import Reasoner
```

```python
OFL_VERSIONEN  = REPO_ROOT.parent / 'OFL Versionen'
ONTOLOGIES_DIR = REPO_ROOT / 'ontologies'

#TBOX_IN  = str(REPO_ROOT / 'ontologies' / '.ttl')
FMA_FILE = str(ONTOLOGIES_DIR / 'fma_obo.owl')
BFO_FILE = str(ONTOLOGIES_DIR / 'bfo-core.owl')
RO_FILE  = str(ONTOLOGIES_DIR / 'ro.owl')
OBI_FILE = str(ONTOLOGIES_DIR / 'obi.owl')
COB_FILE = str(ONTOLOGIES_DIR / 'cob.owl')

TBOX_OUT = str(ONTOLOGIES_DIR / 'ofl_2.1.0_for_inference.owl')
TBOX_INF = str(ONTOLOGIES_DIR / 'flap_onto_with_variants_3_inferred.owl')

print('Checking input files...')
missing = [path for path in (TBOX_IN, FMA_FILE, BFO_FILE, RO_FILE, OBI_FILE, COB_FILE) if not Path(path).exists()]
if missing:
    raise FileNotFoundError(f'Missing required files: {missing}')
print('OK')
```
