# Run pipeline

This notebook runs the flap variant generation pipeline using `wound_ontology` package classes and the native `owlready2` reasoner.

```python
from __future__ import annotations
import os
import sys
from pathlib import Path

# Ensure the wound_ontology package can be imported from the sibling workspace.
NOTEBOOK_DIR = Path.cwd()
PACKAGE_ROOT = NOTEBOOK_DIR.parents[1] / 'Wound ontology' / 'src'
if str(PACKAGE_ROOT) not in sys.path:
    sys.path.insert(0, str(PACKAGE_ROOT))

from wound_ontology.export.owl_variant_generator import OWLVariantGenerator
from wound_ontology.export.tbox import TBox
from wound_ontology.export.reasoner import Reasoner
```

```python
TBOX_IN = 'flap_onto_with_variants_2.owl'
FMA_FILE = 'fma_obo.owl'
TBOX_OUT = 'flap_onto_with_variants_3.owl'
TBOX_INF = 'flap_onto_with_variants_3_inferred.owl'

print('Checking input files...')
missing = [path for path in (TBOX_IN, FMA_FILE) if not Path(path).exists()]
if missing:
    raise FileNotFoundError(f'Missing required files: {missing}')
print('OK')
```

```python
print('Step 1: Generate variant subclasses')
tbox_onto = TBox().load(TBOX_IN)
fma_onto = TBox().load(FMA_FILE)
gen = OWLVariantGenerator(tbox_onto, fma_onto)
gen.generate()
gen.save(TBOX_OUT)
print(f'Created {len(gen.created)} variant subclasses.')
print(f'Saved {TBOX_OUT}')
```

```python
print('Step 2: Run native owlready2 reasoner')
reasoner = Reasoner()
reasoner.reason(input_file=TBOX_OUT, output_file=TBOX_INF)
print(f'Saved {TBOX_INF}')
```
